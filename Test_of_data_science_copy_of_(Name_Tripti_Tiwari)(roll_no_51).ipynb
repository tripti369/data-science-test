```python

import pandas as pd
```

```python


import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Month': ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Temperature': [5, 7, 12, 18, 24, 28, 30, 29, 25, 18, 12, 7]
})

plt.plot(data['Month'], data['Temperature'], marker='o', color='blue')
plt.title("Monthly Temperature Trend")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
# answer is line chart
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = pd.DataFrame({
    'Month': ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Temperature': [5, 7, 12, 18, 24, 28, 30, 29, 25, 18, 12, 7]
})


sns.lineplot(x="Month", y="Temperature", data=data, marker="o")
plt.title("Monthly Temperature Trend")
plt.show()
```

```python

data = pd.DataFrame({
    'Product': ['A','B','C','D','E'],
    'Sales': [120, 90, 150, 60, 110]
})

plt.bar(data['Product'], data['Sales'], color='green')
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.show()
```

```python

sns.barplot(x="Product", y="Sales", data=data, palette="viridis")
plt.title("Sales by Product")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Data
data = pd.DataFrame({
    'Source': ['Coal', 'Solar', 'Hydro', 'Wind'],
    'Percentage': [45, 20, 25, 10]
})

# Plot
plt.pie(data['Percentage'], labels=data['Source'], autopct='%1.1f%%', startangle=90)
plt.title("Energy Consumption Types")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Data
data = pd.DataFrame({
    'Height': [150, 155, 160, 165, 170, 175],
    'Weight': [45, 50, 55, 60, 65, 70]
})

# Plot
plt.scatter(data['Height'], data['Weight'], color='blue')
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight of Students")
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data
data = pd.DataFrame({
    'Age': [18, 20, 22, 25, 30, 35],
    'Income': [15000, 20000, 25000, 30000, 50000, 70000],
    'Spending Score': [30, 40, 50, 60, 70, 80],
    'Savings': [5000, 8000, 10000, 15000, 25000, 40000]
})

# Correlation Matrix
corr = data.corr()

# Plot
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data
data = pd.DataFrame({
    'Study Hours': [2, 3, 4, 5, 6, 7],
    'Marks': [40, 50, 55, 65, 70, 75],
    'Height': [150, 155, 160, 165, 170, 175],
    'Weight': [45, 50, 55, 60, 65, 70]
})

# Plot
sns.pairplot(data)
plt.suptitle("Pair Plot of Student Features", y=1.02)
plt.show()
```
